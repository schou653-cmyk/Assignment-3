# DATA 2402: Assignment 3 — F1 Tire Strategy Simulation
This notebook imports `RaceSimulator`, `plot_tires`, and `find_best_strategy` from `part 2.py` and demonstrates normal, special/edge, and error test cases.



```python
import importlib
import numpy as np
import pandas as pd

# Import from "part 2.py" (handles the space in the filename cleanly)
part2 = importlib.import_module("part 2")
RaceSimulator = part2.RaceSimulator
plot_tires = part2.plot_tires
find_best_strategy = part2.find_best_strategy

sim = RaceSimulator()
sim.tires  # Display the tire parameters DataFrame
```

```python
# Verify degradation curves match the assignment specification
plot_tires(sim)
```

```python
# Test Case 1: Matches Excel template default (Pit stops at laps 15 & 23 -> stints 14, 8, 38)
# Expected result from Part 1 Excel sheet: ~173.8884s
t1 = sim.simulate(stint_lengths=[14, 8, 38], tire_types=["Soft", "Medium", "Hard"])
print(f"Normal Test 1 ([14, 8, 38] | Soft, Medium, Hard): {t1:.4f}s (Expected: ~173.8884s)")

# Test Case 2: Balanced 20-lap stints (Pit stops at laps 21 & 41 -> stints 20, 20, 20)
# Expected result from PDF example: ~168.30s
t2 = sim.simulate(stint_lengths=[20, 20, 20], tire_types=["Soft", "Medium", "Hard"])
print(f"Normal Test 2 ([20, 20, 20] | Soft, Medium, Hard): {t2:.4f}s (Expected: ~168.30s)")
```

```python
# Edge Case 1: Minimum possible stint lengths (1-lap stints)
t_edge1 = sim.simulate(stint_lengths=[1, 1, 58], tire_types=["Soft", "Hard", "Hard"])
print(f"Edge Test 1 ([1, 1, 58] | Soft, Hard, Hard): {t_edge1:.4f}s")

# Edge Case 2: Using only 2 distinct tire types across 3 stints
t_edge2 = sim.simulate(stint_lengths=[25, 25, 10], tire_types=["Medium", "Medium", "Soft"])
print(f"Edge Test 2 ([25, 25, 10] | Medium, Medium, Soft): {t_edge2:.4f}s")
```

```python
error_cases = [
    ("Only 1 tire type used", [20, 20, 20], ["Soft", "Soft", "Soft"]),
    ("Stints do not sum to 60", [15, 20, 20], ["Soft", "Medium", "Hard"]),
    ("Negative or zero stint length", [0, 30, 30], ["Soft", "Medium", "Hard"]),
    ("Wrong number of stints", [30, 30], ["Soft", "Hard"]),
    ("Unknown tire compound", [20, 20, 20], ["Soft", "UltraSoft", "Hard"]),
]

for description, stints, tires in error_cases:
    try:
        sim.simulate(stints, tires)
        print(f"FAIL ({description}): No error raised!")
    except ValueError as e:
        print(f"PASS ({description}) -> Caught expected ValueError: {e}")
```

```python
best_stints, best_tires, min_time = find_best_strategy(sim)
pit_1 = best_stints[0] + 1
pit_2 = best_stints[0] + best_stints[1] + 1

print("=== Optimal F1 Race Strategy ===")
print(f"Stint Lengths : {best_stints}")
print(f"Pit Stop Laps : Lap {pit_1} and Lap {pit_2}")
print(f"Tire Compounds: {best_tires}")
print(f"Min Time Added: {min_time:.4f} seconds")
```
